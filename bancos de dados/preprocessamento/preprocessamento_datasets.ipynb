```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
from collections import Counter
```

```python
nome_xls = 'excel'
arquivos_xls = 1
nome_df = 'dataset_cigarro'

df_total = pd.concat([
    pd.read_excel(f"{nome_xls}{i}.xls")
    for i in range(1, arquivos_xls + 1)
], ignore_index=True)

df_total = df_total.dropna(subset='DOI')
display(df_total)
```

```python
df_scopus = pd.read_csv('scopus.csv')
df_scopus
```

```python
abstracts = list(df_total["Abstract"]) + list(df_scopus["Abstract"])
titulos = list(df_total["Article Title"]) + list(df_scopus["Title"])
dois = list(df_total["DOI"]) + list(df_scopus["DOI"])
autores = list(df_total["Authors"]) + list(df_scopus["Authors"])
anos = list(df_total["Publication Year"]) + list(df_scopus["Year"])
wos_id = wos_id = list(df_total["UT (Unique WOS ID)"]) + list(df_scopus["EID"])
artigo_id = []

for i in range(len(abstracts)):
    a = i + 1
    artigo_id.append(a)

resultados = {
    "title": titulos,
    "authors": autores,
    "abstract": abstracts,
    "DOI": dois,
    "year": anos,
    "article_id": artigo_id,
    "wos_id": wos_id
}

df = pd.DataFrame(resultados)
```

```python
len(df)
```

```python
df = df.drop_duplicates(subset=["DOI"])
df = df.drop_duplicates(subset=["title"])
```

```python
len(df)
```

```python
df.dropna(subset=["DOI"], inplace=True)
```

```python
df.to_excel(f"{nome_df}.xlsx", index=False)
print(f"Saved '{nome_df}.xlsx'.")
```
